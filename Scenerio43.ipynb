```python
data1 = [(101, 'John'),
(102, 'David'),
(103, 'Emma'),
(104, 'Sara'),
(105, 'Mike'),
(106, 'Chris')]

schema1 = ["customer_id","customer_name"]

customersdf = spark.createDataFrame(data1,schema1)
display(customersdf)
```

```python
data2 = [(1, 101, 500),
(2, 102, 300),
(3, 101, 700),
(4, 104, 200),
(5, 106, 900)]

schema2 = ["order_id","customer_id","order_amount"]

ordersdf = spark.createDataFrame(data2,schema2)
display(ordersdf)
```

```python
joindf = customersdf.join(ordersdf, customersdf.customer_id == ordersdf.customer_id, "left")
display(joindf.select(customersdf.customer_id,customersdf.customer_name,ordersdf.order_id,ordersdf.order_amount))
```

```python
from pyspark.sql import * 
from pyspark.sql.types import *
from pyspark.sql.functions import *


finaldf = joindf.filter(col("order_id").isNull())
display(finaldf.select(customersdf.customer_id,customersdf.customer_name))
```
