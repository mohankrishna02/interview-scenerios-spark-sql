```python
from pyspark.sql import *
from pyspark.sql.types import *
from pyspark.sql.functions import *

data = [(101, 'John', 'Dallas',   '2024-01-01 10:00:00'),
(101, 'John', 'Austin',   '2024-01-05 09:00:00'),
(102, 'David', 'Chicago', '2024-01-02 11:00:00'),
(102, 'David', 'Boston',  '2024-01-06 15:00:00'),
(103, 'Emma', 'Seattle',  '2024-01-03 08:00:00')]

columns = ["customer_id","customer_name","city","updated_at"]

df = spark.createDataFrame(data,columns)
display(df)
```

```python
windowdf = Window.partitionBy("customer_id").orderBy(col("updated_at").desc())

rankdf = df.withColumn("rn", row_number().over(windowdf))
display(rankdf)
```

```python
finaldf = rankdf.filter(col("rn")==1).drop("rn")
display(finaldf)
```
