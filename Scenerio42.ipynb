```python
from pyspark.sql import *
from pyspark.sql.types import *
from pyspark.sql.functions import *

data = [(1001, 101, 500.00, '2024-01-01'),
(1002, 102, 300.00, '2024-01-02'),
(1001, 101, 500.00, '2024-01-01'),
(1003, 103, 700.00, '2024-01-03'),
(1004, 104, 200.00, '2024-01-04'),
(1002, 102, 300.00, '2024-01-02'),
(1002, 102, 300.00, '2024-01-02'),
(1005, 105, 900.00, '2024-01-05')]

columns = ["transaction_id","customer_id","amount","transaction_date"]

df = spark.createDataFrame(data,columns)
display(df)
```

```python
countdf = df.groupBy("transaction_id").agg((count("transaction_id")).alias("occurence_count"))
display(countdf)
```

```python
finaldf = countdf.filter(col("occurence_count")>1)
display(finaldf)
```
